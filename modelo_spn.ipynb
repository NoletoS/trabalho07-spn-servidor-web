# Servidor web de capacidade finita
Avaliação estacionária de uma GSPN com chegada Poisson, serviço exponencial e bloqueio na admissão. As marcas de tempo são dadas em segundos.

## Estados e disparos
A marcação ordena **entrada, fila, ocupados, livres, vagas**. A admissão consome uma vaga, o início de serviço consome um processador livre e a conclusão devolve ambos. Uma chegada com zero vagas é descartada. O programa enumera as marcações tangíveis e calcula probabilidades estacionárias por balanço local.

```python
from avaliar import evaluate, run
resultados = run()
print(f"{len(resultados)} cenários avaliados. Arquivos gerados: resultados.csv e estacionario.csv.")
```

## Métricas
TP é a taxa de conclusões, MRT é a resposta média, MQT é a espera média, U é a utilização, DP é a probabilidade de descarte e MPT é o serviço médio. A resposta é condicionada à admissão.

```python
from IPython.display import HTML, display
campos = ("id", "lambda", "c", "B", "K", "TP", "MRT_ms", "MQT_ms", "U_pct", "DP_pct", "MPT_ms")
head = "".join(f"<th>{k}</th>" for k in campos)
rows = []
for linha in resultados:
    valores = [f"{linha[k]:.4f}" if isinstance(linha[k], float) else str(linha[k]) for k in campos]
    rows.append("<tr>" + "".join(f"<td>{v}</td>" for v in valores) + "</tr>")
display(HTML("<table><thead><tr>" + head + "</tr></thead><tbody>" + "".join(rows) + "</tbody></table>"))
```

## Seleção pelo objetivo de desempenho
O estudo adota, para comparação, descarte de até 1% e resposta média de até 30 ms. O resultado vale para os cenários ensaiados.

```python
for linha in resultados:
    if linha["DP_pct"] <= 1 and linha["MRT_ms"] <= 30:
        print(linha["id"], "atende às duas metas")
```
